###**WordNet**
WordNet is a lexical database for the English language that groups words into sets of synonyms called **synsets**. It also stores relationships between words such as synonym, antonym, hypernym (general term), and hyponym (specific term). WordNet is widely used in **natural language processing** to understand word meanings and relationships.

###**Word Similarity on English Text**
Word similarity refers to measuring how closely related two words are in meaning. It is used in text analysis to identify words that have similar meanings or contexts. Techniques such as semantic analysis, WordNet relationships, or word embeddings are commonly used to compute similarity between words.

```python
from nltk.corpus import wordnet
antonyms = []
# defining the function to find Antonyms
def TofindAntonyms(x):
  for syn in wordnet.synsets(x):
    for lm in syn.lemmas():
      if lm.antonyms():
        antonyms.append(lm.antonyms()[0].name()) #adding into antonyms
  return antonyms
print(set(TofindAntonyms("bright")))
```

```python
print(set(TofindAntonyms("inactive")))
```

```python
print(set(TofindAntonyms("good")))
```

```python
synonyms=[]
antonyms=[]
for syn in wordnet.synsets("good"):
  for l in syn.lemmas():
    synonyms.append(l.name())
    if l.antonyms():
      antonyms.append(l.antonyms()[0].name())
print("Synonyms: ",set(synonyms))
print("Antonyms: ",set(antonyms))
```

```python
# WordSimilarity
car = wordnet.synset('car.n.01')
automobile = wordnet.synset('automobile.n.01')
print("Similarity between car and automobile", car.path_similarity(automobile))
```

# **Wu-Palmer Similarity**

It calculates relatedness by considering the depths of the two synsets in the WordNet taxonomies, along with the depth of the LCS (Least Common Subsumer). The score can be 0 < score <= 1. The score can never be zero because the depth of the LCS is never zero (the depth of the root of taxonomy is one). It calculates the similarity based on how similar the word senses are and where the Synsets occur relative to each other in the hypernym tree


```python
# Wu-Palmer Similarity
from nltk.corpus import wordnet as wn
w1 = wordnet.synset('run.v.01') # v here denotes the tag verb
w2 = wordnet.synset('sprint.v.01')
print("Wu-Palmer Similarity between run and sprint", w1.wup_similarity(w2))
```

```python
jump = wn.synset('jump.v.01')
leap = wn.synset('leap.v.01')
ship = wn.synset('ship.n.01')
print("Wu-Palmer Similarity between jump and leap", jump.wup_similarity(leap))
```
