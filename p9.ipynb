###**Word2Vec on Wikipedia Articles and Word Similarity**
Word2Vec is a Natural Language Processing technique that converts words into numerical vectors by learning from large text datasets such as Wikipedia articles. Words that appear in similar contexts get similar vector representations. The similarity between words is then calculated (often using cosine similarity) to determine how closely related their meanings are.

```python
from nltk.tokenize import sent_tokenize
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

import string
import warnings
from gensim.models import Word2Vec
warnings.filterwarnings(action='ignore', category=UserWarning, module='gensim')

import wikipedia
from wikipedia import search
from wikipedia import page

titles=search("Data Science")
wikipage=page(titles[0])
wikipage.content
wikipage.categories
wikipedia.summary("Data Science", sentences=1)
wikipedia.search("Data Science")
```

```python
def preprocessing(text):
  result=[]
  sent=sent_tokenize(text)
  for sentence in sent:
    words=word_tokenize(sentence)
    tokens=[w for w in words if w.lower() not in string.punctuation]
    stopw= stopwords.words('english')
    tokens=[token for token in tokens if token.lower() not in stopw]
    tokens=[word for word in tokens if len(word)>=3]
    Lemma=WordNetLemmatizer()
    tokens=[Lemma.lemmatize(word) for word in tokens]
    result+=[tokens]
  return result
```

```python
text_p= preprocessing(wikipage.content)
text_p[0]
```

```python
# Define your parameters
min_count=2
vector_size=50
window=4
wikimodel=Word2Vec(text_p,min_count=min_count, vector_size=vector_size,window=window)
```

```python
vocab=list(wikimodel.wv.index_to_key)
vocab[:20]
```

```python
wikimodel.wv.most_similar(positive = ["analysis", "field"], topn = 3)
print(wikimodel.wv.similarity("data", "information"))
print(wikimodel.wv.similarity("data", "science"))
print(wikimodel.wv.similarity("data", "computing"))
```

```python
titles = search("statistic")
wikipage = page(titles[0])
wikipage.content
wikipage.categories
wikipedia.summary("statistic", sentences = 1)
wikipedia.search("statistic")
```

```python
def preprocessing(text):
    result = []
    sent = sent_tokenize(text)
    for sentence in sent:
        words = word_tokenize(sentence)
        tokens = [w for w in words if w.lower() not in string.punctuation]
        stopw = stopwords.words("english")
        tokens = [token for token in tokens if token not in stopw]
        tokens = [word for word in tokens if len(word)>= 3]
        lemma = WordNetLemmatizer()
        tokens = [lemma.lemmatize(word) for word in tokens]
        result += [tokens]
    return result
text_p = preprocessing(wikipage.content)
text_p[0]
```

```python
# Define your parameters
min_count = 2
vector_size = 50
window = 4
# Create the Word2Vec model
wikimodel = Word2Vec(text_p, min_count=min_count, vector_size=vector_size, window=window)
```

```python
#Get the vocabulary
vocab = list(wikimodel.wv.index_to_key)
vocab[:20]
```

```python
wikimodel.wv.most_similar(positive = ["statistical", "hypothesis"], topn = 3)
print(wikimodel.wv.similarity("probability", "sample"))
print(wikimodel.wv.similarity("probability", "value"))
print(wikimodel.wv.similarity("data", "error"))
```
