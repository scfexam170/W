###**Word Cloud**
A **word cloud** is a visual representation of text data where words appear in different sizes based on how frequently they occur in the text. Frequently used words appear **larger**, while less frequent words appear **smaller**.
It is commonly used to **identify important keywords, themes, or trends** in large text datasets such as survey responses, articles, or social media posts.


```python
from wordcloud import WordCloud
```

```python
import matplotlib.pyplot as plt
import matplotlib.pyplot as pPlt
```

```python
from wordcloud import WordCloud, STOPWORDS
import numpy as npy
```

```python
from PIL import Image
```

```python
dataset = open("/content/Indian Motorcycle.txt", "r").read()
dataset = dataset.upper()
```

```python
maskArray = npy.array(Image.open("/content/video-300x300.jpg"))
cloud = WordCloud(background_color = "#000012", min_font_size=5, colormap='Blues_r',
max_font_size=70, font_path='/content/LEMONMILK-Medium.otf', max_words= 2000, width=1400, height=800,
collocations=True, mask = maskArray, stopwords = set(STOPWORDS)).generate(dataset)
```

```python
# plot the WordCloud image
plt.figure(figsize = (6, 10), facecolor = None)
plt.imshow(cloud)
plt.axis("off")
```
