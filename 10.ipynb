###**Entity Recognition(NER) & Chunking**

Named Entity Recognition (NER) and Chunking are Natural Language Processing (NLP) techniques used to extract structured information from unstructured text. NER identifies and categorizes key entities (people, places, organizations, times), while chunking groups words into meaningful phrases (noun phrases, verb phrases) based on syntactic roles

```python
from nltk.tokenize import word_tokenize
from nltk.tag import pos_tag
from nltk.chunk import ne_chunk

dataset = """Abraham Lincoln was an American statesman, lawyer, and the 16th president of the United States, serving from 1861 until his assassination in 1865."""

dataset_tag = pos_tag(word_tokenize(dataset))
print(dataset_tag)
```

```python
# Apply NER with ne_chunk
data_ner=ne_chunk(dataset_tag)
print(data_ner)
```

```python
# Chunking
import nltk
from nltk.tokenize import word_tokenize
from nltk.tag import pos_tag
from nltk.chunk import RegexpParser

# Tokenize the data
dataset = """Taj Mahal is one of the world's most celebrated structures in the world. It is a stunning white marble mausoleum located in Agra, India."""

new_data = word_tokenize(dataset)
print(new_data)
```

```python
# Apply the POS Tagging
postagging = pos_tag(new_data)
print(postagging)
```

```python
# defining sequence of chunk
sequence_chunk = """
chunk:
{<NNPS>+}
{<NNP>+}
{<NN>+}
"""
# creating object with regular expression
chunk=RegexpParser(sequence_chunk)
chunk_result=chunk.parse(postagging)
print(chunk_result)
```
