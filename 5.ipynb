###**Naive Bayes Classifier**
Naive Bayes Classifier is a **supervised machine learning algorithm** used for classification tasks. It is based on **Bayes’ theorem** and assumes that the features are **independent of each other**. It is commonly used in applications such as **text classification, spam detection, and sentiment analysis** because it works efficiently with large datasets and text data.

```python
import nltk
```

```python
nltk.download('all')
```

```python
import random

from nltk.corpus import movie_reviews
```

```python
documents = []

for category in movie_reviews.categories():
    for fileid in movie_reviews.fileids(category):
        documents.append((list(movie_reviews.words(fileid)), category))

random.shuffle(documents)

print(documents[1])
```

```python
#Normalize the dataset
all_words = []
for w in movie_reviews.words():
    all_words.append(w.lower())

all_words = nltk.FreqDist(all_words)
print(all_words.most_common(15))
print(all_words["love"])
```

```python
word_features = list(all_words.keys())[:3000]

def find_features(document):
    words = set(document)
    features = {}
    for w in word_features:
      features[w] = (w in words)
    return features

print((find_features(movie_reviews.words('neg/cv000_29416.txt'))))
featuresets = [(find_features(rev), category)
               for (rev, category) in documents]
```

```python
training_set=featuresets[:1900]
testing_set=featuresets[1900:]
```

```python
classifier=nltk.NaiveBayesClassifier.train(training_set)
print("Accuracy:",(nltk.classify.accuracy(classifier, testing_set))*100)
```
